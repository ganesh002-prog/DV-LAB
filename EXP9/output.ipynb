```python

import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
data = pd.read_csv("city.csv")

# 1. Bar chart: Sales by Category
data.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("BarChart.png")
plt.show()
plt.close()

# 2. Pie chart: Sales by Region
data.groupby("Region")["Sales"].sum().plot(
    kind="pie", autopct="%1.1f%%"
)
plt.title("Sales by Region")
plt.ylabel("")
plt.tight_layout()
plt.savefig("PieChart.png")
plt.show()
plt.close()

# 3. Line graph: Sales vs Quantity
plt.plot(data["Quantity"], data["Sales"], marker="o")
plt.title("Sales vs Quantity")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.tight_layout()
plt.savefig("SalesVsQuantity.png")
plt.show()
plt.close()

print("All three charts created successfully!")
```
