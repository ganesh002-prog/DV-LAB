```r
x <- c(1,2,3,4,5)
y <- c(10,15,12,18,20)
plot(x, y,
     type="o",
     col="blue",
     main="Line Graph",
     xlab="X Values",
     ylab="Y Values")
```

```r
marks <- c(75,85,60,90,80)
barplot(marks,
        names.arg=c("Tamil","English","Maths","Science","Computer"),
        col="green",
        main="Subject Marks",
        xlab="Subjects",
        ylab="Marks")
```

```r
values <- c(30,25,20,15,10)
labels <- c("Tamil","English","Maths","Science","Computer")
pie(values,
    labels=labels,
    col=rainbow(5),
    main="Subject Distribution")
```

```r
x <- c(2, 4, 6, 8, 10)
y <- c(5, 9, 7, 12, 15)

plot(x, y,
     col = "red",
     pch = 19,
     main = "Scatter Plot",
     xlab = "X Values",
     ylab = "Y Values")
```

```r
marks <- c(45, 55, 60, 65, 70, 75, 80, 85, 90, 95)
hist(marks,
     col = "orange",
     main = "Distribution of Marks",
     xlab = "Marks",
     ylab = "Frequency")
```
