```python

import pandas as pd
import matplotlib.pyplot as plt

# Load the uploaded CSV file
data = pd.read_csv("sales.csv")

# Check required columns
print("Columns:", data.columns.tolist())

# Bar chart: Sales by Category
sales = data.groupby("Category")["Sales"].sum()

sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("BarChart.png")
plt.show()
plt.close()

# Pie chart: Category Distribution
sales.plot(kind="pie", autopct="%1.1f%%")
plt.title("Category Distribution")
plt.ylabel("")
plt.tight_layout()
plt.savefig("PieChart.png")
plt.show()
plt.close()

# Line graph: Sales vs Quantity
plt.plot(data["Quantity"], data["Sales"], marker="o")
plt.title("Sales vs Quantity")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.tight_layout()
plt.savefig("SalesVsQuantity.png")
plt.show()
plt.close()

print("All three charts created successfully.")
```
